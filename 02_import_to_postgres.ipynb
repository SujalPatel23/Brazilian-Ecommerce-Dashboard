```python
import pandas as pd
from sqlalchemy import create_engine
```

```python
orders = pd.read_csv("orders_cleaned.csv")
items = pd.read_csv("items_cleaned.csv")
customers = pd.read_csv("customers_cleaned.csv")
payments = pd.read_csv("payements_cleaned.csv")
products = pd.read_csv("products_cleaned.csv")
```

```python
engine = create_engine("postgresql+psycopg2://postgres:1234@localhost:5432/olist_project")
```

```python
orders.to_sql("orders", engine, if_exists = "replace", index=False)
items.to_sql("items", engine, if_exists = "replace", index=False)
customers.to_sql("customers",engine,if_exists = "replace", index=False)
payments.to_sql("payments",engine,if_exists = "replace", index=False)
products.to_sql("products",engine,if_exists = "replace", index=False)
```
